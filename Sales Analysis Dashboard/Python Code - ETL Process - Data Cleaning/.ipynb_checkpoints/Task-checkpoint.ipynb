```python
import pandas as pd

# Load your dataset
df = pd.read_excel(r"C:\Users\asus\OneDrive\Desktop\Task\Sales Data.xlsx")
print("start reading")
# Replace blanks ("") with NaN so pandas can detect them
df = df.replace(r'^\s*$', pd.NA, regex=True)

# Forward fill specific columns
cols_to_fill = ["order_id", "date format", "city", "branch", "product_category"]
df[cols_to_fill] = df[cols_to_fill].ffill()

# Save cleaned dataset
df.to_excel("orders_cleaned.xlsx", index=False)

print("✅ Nulls replaced with previous values successfully!")
```

```python
# 1️⃣ Date Dimension
# -------------------------
date_dim = df[["date format"]].drop_duplicates().reset_index(drop=True)
date_dim["date format"] = pd.to_datetime(date_dim["date format"], errors="coerce")

date_dim["Date"] = date_dim["date format"].dt.date
date_dim["Year"] = date_dim["date format"].dt.year
date_dim["Month"] = date_dim["date format"].dt.month
date_dim["Day"] = date_dim["date format"].dt.day
date_dim["Weekday"] = date_dim["date format"].dt.day_name()

# Add surrogate key
date_dim.insert(0, "Date_ID", range(1, 1 + len(date_dim)))

# -------------------------
# 2️⃣ Branch Dimension
# -------------------------
branch_dim = df[["branch", "city"]].drop_duplicates().reset_index(drop=True)
branch_dim.rename(columns={"branch": "Branch_Name", "city": "City"}, inplace=True)

# Add surrogate key
branch_dim.insert(0, "Branch_ID", range(100, 100 + len(branch_dim)))

# -------------------------
# 3️⃣ Product Dimension
# -------------------------
product_dim = df[["product name", "product_category"]].drop_duplicates().reset_index(drop=True)
product_dim.rename(columns={
    "product name": "Product_Name",
    "product_category": "Product_Category"
}, inplace=True)

# Add surrogate key (Product_ID)
product_dim.insert(0, "Product_ID", range(1000, 1000 + len(product_dim)))

# -------------------------
# Save to Excel (each dim in separate sheet)
# -------------------------
with pd.ExcelWriter("DWH_Dimensions.xlsx") as writer:
    date_dim.to_excel(writer, sheet_name="Date_Dim", index=False)
    branch_dim.to_excel(writer, sheet_name="Branch_Dim", index=False)
    product_dim.to_excel(writer, sheet_name="Product_Dim", index=False)

print("✅ Dimensions created successfully! Saved to DWH_Dimensions.xlsx")
```

```python
# 1️⃣ Date Dimension
# -------------------------
date_dim = df[["date format"]].drop_duplicates().reset_index(drop=True)
date_dim["date format"] = pd.to_datetime(date_dim["date format"], errors="coerce")

date_dim["Full_Date"] = date_dim["date format"].dt.strftime("%Y-%m-%d")
date_dim["Year"] = date_dim["date format"].dt.year
date_dim["Month"] = date_dim["date format"].dt.month
date_dim["Day"] = date_dim["date format"].dt.day
date_dim["Weekday"] = date_dim["date format"].dt.day_name()

date_dim.insert(0, "Date_ID", range(1000000, 1000000 + len(date_dim)))

# -------------------------
# 2️⃣ Branch Dimension
# -------------------------
branch_dim = df[["branch", "city"]].drop_duplicates().reset_index(drop=True)
branch_dim.rename(columns={"branch": "Branch_Name", "city": "City"}, inplace=True)

branch_dim.insert(0, "Branch_ID", range(100, 100 + len(branch_dim)))

# -------------------------
# 3️⃣ Product Dimension
# -------------------------
product_dim = df[["product name", "product_category"]].drop_duplicates().reset_index(drop=True)
product_dim.rename(columns={
    "product name": "Product_Name",
    "product_category": "Product_Category"
}, inplace=True)

product_dim.insert(0, "Product_ID", range(1000, 1000 + len(product_dim)))

# -------------------------
# 4️⃣ Fact Table
# -------------------------
fact_orders = df.copy()

# 🔹 توحيد صيغة التاريخ كنص
fact_orders["date format"] = pd.to_datetime(fact_orders["date format"], errors="coerce")
fact_orders["Full_Date"] = fact_orders["date format"].dt.strftime("%Y-%m-%d")

# ✅ Map Date_ID مباشرة باستخدام dictionary
date_map = dict(zip(date_dim["Full_Date"], date_dim["Date_ID"]))
fact_orders["Date_ID"] = fact_orders["Full_Date"].map(date_map)

# ✅ Map Branch_ID
branch_map = dict(zip(zip(branch_dim["Branch_Name"], branch_dim["City"]), branch_dim["Branch_ID"]))
fact_orders["Branch_ID"] = fact_orders.apply(lambda x: branch_map.get((x["branch"], x["city"])), axis=1)

# ✅ Map Product_ID
product_map = dict(zip(zip(product_dim["Product_Name"], product_dim["Product_Category"]), product_dim["Product_ID"]))
fact_orders["Product_ID"] = fact_orders.apply(lambda x: product_map.get((x["product name"], x["product_category"])), axis=1)

# Keep only IDs + measures
fact_orders = fact_orders[[
    "order_id",
    "Date_ID",
    "Branch_ID",
    "Product_ID",
    "quantity",
    "total",
    "order_total"
]]

# -------------------------
# Save everything
# -------------------------
with pd.ExcelWriter("DWH_Model.xlsx") as writer:
    date_dim.to_excel(writer, sheet_name="Date_Dim", index=False)
    branch_dim.to_excel(writer, sheet_name="Branch_Dim", index=False)
    product_dim.to_excel(writer, sheet_name="Product_Dim", index=False)
    fact_orders.to_excel(writer, sheet_name="Fact_Orders", index=False)

print("✅ Dimensions & Fact Table saved in 'DWH_Model.xlsx'")
```
